## 2. Проверка гипотезы в Python

Гипотеза звучит так: пользователи из Санкт-Петербурга проводят в среднем больше времени за чтением и прослушиванием книг в приложении, чем пользователи из Москвы. Попробуйте статистически это доказать, используя одностороннюю проверку гипотезы с двумя выборками:

- Нулевая гипотеза H₀: Средняя активность пользователей в часах в двух группах (Москва и Санкт-Петербург) не различается.

- Альтернативная гипотеза H₁: Средняя активность пользователей в Санкт-Петербурге больше, и это различие статистически значимо.

<div class="alert alert-block alert-danger">
<h2> Комментарий ревьюера v1<a class="tocSkip"></h2>

<b>На доработку🤔:</b>Тут действительно нужен `t-тест`, однако перед его проведением важно проверить пересечения по группам, если взглянуть по полю `puid`, то мы увидим дубликаты и их лучше сразу убрать

<div class="alert alert-info"> <b>Комментарий студента: исправлено</b></div>

```python
msk = df[df['city'] == 'Москва']['hours']
spb = df[df['city'] == 'Санкт-Петербург']['hours']

print(f'Москва: n={len(msk)}, среднее={msk.mean():.4f}, std={msk.std():.4f}')
print(f'СПб:    n={len(spb)}, среднее={spb.mean():.4f}, std={spb.std():.4f}')

t_stat, p_value = stats.ttest_ind(spb, msk, equal_var=False, alternative='greater')

alpha = 0.05

print(f't-статистика: {t_stat:.4f}')
print(f'p-value:      {p_value:.4f}')
print(f'Уровень значимости α: {alpha}')
print()
if p_value < alpha:
    print('Результат: p-value < α → ОТВЕРГАЕМ нулевую гипотезу.')
    print('Вывод: среднее время активности пользователей СПб статистически значимо БОЛЬШЕ, чем в Москве.')
else:
    print('Результат: p-value >= α → НЕ ОТВЕРГАЕМ нулевую гипотезу.')
    print('Вывод: статистически значимых различий не обнаружено.')
```

Тест **односторонний**: `alternative='greater'` (проверяем, что μ(СПб) > μ(Москвы)).

### Возможные причины результатов

1. **Высокая вариативность данных.** Распределение активности сильно правосторонне скошено: большинство пользователей читают мало, но небольшая группа "суперпользователей" с сотнями часов сильно завышает средние и увеличивает дисперсию. Это снижает мощность теста и затрудняет обнаружение реальных различий между городами.

2. **Поведенческое сходство аудитории.** Москва и Санкт-Петербург — оба крупных мегаполиса с похожей демографией активных пользователей подписных сервисов. Возможно, реальные различия в потреблении контента между ними минимальны, и исходная гипотеза не подтверждается данными.

## 4. Проведите оценку результатов A/B-тестирования:

- Проверьте изменение конверсии подходящим статистическим тестом, учитывая все этапы проверки гипотез.

<div class="alert alert-block alert-danger">
<h2> Комментарий ревьюера v1<a class="tocSkip"></h2>

<b>На доработку🤔:</b> Также, тут действительно нужен z-тест, но он проверяет только статистическую значимость изменений.  То что увеличение есть тобой уже выявлено  выше рассчетами по реальным данным  

Гипотезы в данном случае  правильнее сформулировать примерно так
    
    Уровень конверсии интернет - магазина стат значимо не изменится 
    Уровень конверсии интернет - магазина увеличится с внедрением нового интерфейса сайта и это изменение стат значимо.

```python
# Z-тест для долей (двухвыборочный, односторонний)
# H0: уровень конверсии интернет-магазина статистически значимо не изменится
# H1: уровень конверсии интернет-магазина увеличится с внедрением нового интерфейса, и это изменение статистически значимо

buyers_A = conv_df[conv_df['group'] == 'A']['buyers'].values[0]
buyers_B = conv_df[conv_df['group'] == 'B']['buyers'].values[0]
total_A  = conv_df[conv_df['group'] == 'A']['total'].values[0]
total_B  = conv_df[conv_df['group'] == 'B']['total'].values[0]

count = np.array([buyers_B, buyers_A])
nobs  = np.array([total_B,  total_A])

z_stat, p_value = proportions_ztest(count, nobs, alternative='larger')

print('Z-тест для долей (двухвыборочный, односторонний)')
print(f'Конверсия A: {buyers_A}/{total_A} = {buyers_A/total_A:.4f} ({buyers_A/total_A*100:.2f}%)')
print(f'Конверсия B: {buyers_B}/{total_B} = {buyers_B/total_B:.4f} ({buyers_B/total_B*100:.2f}%)')
print(f'z-статистика: {z_stat:.4f}')
print(f'p-value:      {p_value:.4f}')
print(f'Уровень значимости α: {alpha}')
print()
if p_value < alpha:
    print('p-value < α → ОТВЕРГАЕМ H0')
    print('Конверсия в группе B статистически значимо выше, чем в группе A.')
else:
    print('p-value >= α → НЕ отвергаем H0')
    print('Статистически значимых различий не обнаружено.')
```

<div class="alert alert-info"> <b>Комментарий студента:</b> Скорректировал формулировки гипотез: H0 — конверсия статистически значимо не изменится; H1 — конверсия увеличится с внедрением нового интерфейса, и это изменение статистически значимо. Сам тест и его параметры оставлены без изменений.</div>

### Метод
Использован **двухвыборочный z-тест для долей** (односторонний, `alternative='larger'`):

- Уровень значимости: **α = 0.05**

Z-тест выбран как стандартный метод сравнения конверсий при больших выборках (n > 4900 в каждой группе).
